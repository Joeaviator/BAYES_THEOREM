```python
W_Strep_Throat=0.20
NA_Strep_Throat=0.80
P_W_Strep_Throat=0.85
N_W_Strep_Throat=0.15
N_NA_Strep_Throat=0.98
P_NA_Strep_Throat=0.02
#Probaility that they are Positive and have strep throat
# P_Pos=P_W_Strep_Throat*P_NA_Strep_Throat
# print(f"{P_Pos:.2f}")
Retry=(W_Strep_Throat*P_W_Strep_Throat)+(P_NA_Strep_Throat*NA_Strep_Throat)#All the possible outcomes
P_Pos=(W_Strep_Throat*P_W_Strep_Throat)/Retry
print(f"{P_Pos:.2f}")
```
